```python
from pathlib import Path
# Specify the folder to be plotted
root_path=Path('../example_local/N=200, D_0=0.01, J=1.0, v_0=0.5, aligned_init=1, delta_t=0.5, dt=0.01')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import json
import os

# --- Load Data ---
def load_txt(filename):
    data = np.loadtxt(filename, skiprows=0)
    return data.T[:, :-1]  

x = load_txt(root_path / "x.txt")
y = load_txt(root_path / "y.txt")
v_x = load_txt(root_path / "v_x.txt")
v_y = load_txt(root_path / "v_y.txt")

## Reading input parameters of the simulation
with open(os.path.join(root_path, 'input.json'), 'r') as f:
    params = json.load(f)

N = params.get('N')
L_box_x = params.get('L_box_x')
L_box_y = params.get('L_box_y')
v_0 = params.get('v_0')
dt = params.get('dt')
interval = params.get('interval')
Obs_time_steps = params.get('Obs_time_steps')
```

```python
# Movie with quiver plot
from IPython.display import clear_output

time_steps=np.array(x).shape[1]
k_interval =1
k_range = np.arange(0, time_steps)

for k in k_range:
    if k % k_interval != 0:
        continue
    clear_output(wait=True)
    fig1=plt.figure(figsize=(5,5))
    plt.clf()
    q1=plt.quiver(x[:, k], y[:, k], v_x[:, k], v_y[:, k], color='blue')
    plt.axis('square')
    plt.xlim(-L_box_x/2, L_box_x/2)
    plt.ylim(-L_box_y/2, L_box_y/2)
    plt.title(f'Time step: {k}')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.show()

```

```python
## Animation with quiver plot
import matplotlib.pyplot as plt
import matplotlib.animation as animation

fig, ax = plt.subplots()
quiver = ax.quiver(x[:, 0], y[:, 0], v_x[:, 0], v_y[:, 0], color='blue')
ax.set_xlim(-L_box_x/2, L_box_x/2)
ax.set_ylim(-L_box_y/2, L_box_y/2)
ax.set_aspect('equal')
ax.set_xlabel('x')
ax.set_ylabel('y')

def update(k):
    quiver.set_UVC(v_x[:, k], v_y[:, k])
    quiver.set_offsets(np.c_[x[:, k], y[:, k]])
    ax.set_title(f'Time step: {k}')
    return quiver,

ani = animation.FuncAnimation(fig, update, frames=x.shape[1], interval=50, blit=True)

from IPython.display import HTML
HTML(ani.to_jshtml())
```
